```python
class Car():
    def move(self):
        print("car is moving")
    def stop(self):
        print("car is stopped")
        
```

```python
car_object=Car()  # object of class Car
print(type(car_object))
# check if object belongs to class or not
print(isinstance(car_object,Car))
print(isinstance(car_object,object)) # because object is the parent or root of all class and Car is subclass of of object
# to see hoover on object
```

```python
# call Car methods
car_object.move()
car_object.stop()
```

```python
# list of all attributes of car_object
print(dir(car_object))
```

```python
print(car_object.__dict__)
```

```python
class user:
    def info(self,user_name,email):
        print(f"username is {user_name} and email is {email}")
```

```python
user_obj=user()
user_obj.user_name="pappu" # instance attribute of object user_obj
user_obj.emial="pappuyadav98199@gmail.com"
user_obj.info("pappu","pappuyadav98199@gmail.com")
```

```python
# now lets understand instance attribute 
print(user_obj.__dict__) # here __dict__ store instance attribute of user_obj in key value pairs
```

```python
# another example of instance attributes
user_obj2=user()
user_obj2.user_name="ram"
user_obj2.email="ram@gmail.com"
print(user_obj2.__dict__)
```
