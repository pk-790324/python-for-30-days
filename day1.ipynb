### Q.12 Extract Digit

Calculate:
- The last digit.
- The first digit.
- The sum of the first and last digits.
- Constraint: For extracting the last digit, use the modulus operator. 
- For extracting the first digit, use mathematical operations rather than converting the number into a string.

```python
from math import floor,log10


number=12345

last_digit=number%10
print(f"the last digit is: {last_digit}")

first_digit=floor(number/(10**floor(log10(number))))
print(f"the first digit is: {first_digit}")

print(f"the sum of first and last digit is {first_digit+last_digit}")

```

### Q13. Reverse a three-digit number mathematically

Ask the user for a three-digit number.

Example:

Input: 123

Output: 321


Try to solve it using only:
- Arithmetic operators
- Integer division //
- Modulus %
- Do not convert the number into a string.
- They don't require loops, conditions, functions, or lists yet.

```python
number=int(input("Enter three digit number:"))

last_digit=number%10
first_digit=floor(number/(10**floor(log10(number))))
middle_digit=(number//10)%10

print(f"the given three digit number:{number}")
print(f"the first digit is:{first_digit}")
print(f"the middle digit is: {middle_digit}")
print(f"the last digit is:{last_digit}")

reversed_digit=last_digit*100+middle_digit*10+first_digit

print(f"the reversed digit is:{reversed_digit}")
```
